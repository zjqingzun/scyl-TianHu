## INTRODUCTION

<div align="left">
Các bên tấn công có thể thực hiện quét thăm dò chủ động để thu thập thông tin có thể được sử dụng trong quá trình xác định mục tiêu. Quét chủ động là hình thức kẻ tấn công dò xét cơ sở hạ tầng của nạn nhân thông qua lưu lượng mạng, khác với các hình thức thăm dò khác không bao gồm tương tác trực tiếp.
</div>

Các bên tấn công có thể thực hiện nhiều hình thức quét chủ động khác nhau tùy thuộc vào loại thông tin mà họ muốn thu thập. Các cuộc quét này cũng có thể được thực hiện theo nhiều cách khác nhau, bao gồm việc sử dụng các tính năng sẵn có của giao thức mạng như ICMP. Thông tin từ những cuộc quét này có thể tiết lộ cơ hội cho các hình thức thăm dò khác (ví dụ: Tìm kiếm website/miền mở hoặc Cơ sở dữ liệu kỹ thuật công khai), thiết lập nguồn lực vận hành (ví dụ: Phát triển năng lực hoặc Thu thập năng lực), và/hoặc tiếp cận ban đầu (ví dụ: Dịch vụ từ xa bên ngoài hoặc Khai thác ứng dụng phía máy chủ).
